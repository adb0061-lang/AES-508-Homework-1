```python
import folium
import pandas as pd
```

```python
# used this website for help https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.to_dict.html
csv_data = pd.read_csv('HW1_Data - Sheet1.csv') 
data_list = csv_data.to_dict(orient= 'records')
print(data_list)
```

```python
m = folium.Map(location=[csv_data["Lat"].mean(), csv_data["Long"].mean()], zoom_start=10)

for _, row in csv_data.iterrows():
    folium.Marker(
        location=[row["Lat"], row["Long"]],
        tooltip="Click me!",
        popup=row["Location"],
    ).add_to(m)

m
```
